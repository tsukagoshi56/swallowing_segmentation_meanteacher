# **データベース処理(Googledriveからローカルに落とす)**

```python
import os
import gdown

# 食事行動データベースの設定
# Google スプレッドシート: https://docs.google.com/spreadsheets/d/1AmSBnOUhCBBBjP-9AOESp_lpT3bNK-iK/edit
data_path = "./downloaded_folder/食事行動データベース.xlsx"

# データベース保存先フォルダを作成
os.makedirs(os.path.dirname(data_path), exist_ok=True)

# Google Driveから食事行動データベースをダウンロード
# 注意: 実際のファイルIDは、Googleスプレッドシートをエクスポートした際のIDに置き換える必要があります
database_file_id = "1AmSBnOUhCBBBjP-9AOESp_lpT3bNK-iK"  # スプレッドシートのID

try:
    # スプレッドシートをExcel形式でダウンロード
    url = f"https://docs.google.com/spreadsheets/d/{database_file_id}/export?format=xlsx"
    gdown.download(url, data_path, quiet=False, fuzzy=True)
    print(f"食事行動データベースのダウンロードが完了しました: {data_path}")
except Exception as e:
    print(f"ダウンロードエラー: {e}")
    print("手動でスプレッドシートをExcel形式でダウンロードし、指定のパスに保存してください。")
```

```python
import pandas as pd
data_path = "./downloaded_folder/食事行動データベース.xlsx"
output_folder = "./downloaded_folder/behavior"

#ファイル階層は以下を想定
#
# behavior(output_folder)
#    |
#    |--text_ful
#    |     |--MYN01_ful.txt
#    |      --MCN01_ful.txt
#     --  wav_ful
#          |--MYN01_ful.wav
#           --MCN01_ful.wav
#

database = pd.ExcelFile(data_path)

input_sheet_name = database.sheet_names
#シート名を確認
print(input_sheet_name)
```

```python
#DataFrameとして'若年者（データパス）'を読込み
young_person_df = database.parse(input_sheet_name[3])
#読み込んだシートの先頭10行を表示
young_person_df.head(5)
```

ラベル付け替え

```python
young_person_df= young_person_df.rename(columns={young_person_df.columns[1]: 'カメラ(正面)',
                                                 young_person_df.columns[2]: 'カメラ(上)',
                                                 young_person_df.columns[3]: 'カメラ(横)',
                                                 young_person_df.columns[5]: '加速度(頭)',
                                                 young_person_df.columns[6]: '加速度(左)',
                                                 young_person_df.columns[7]: '加速度(右)',
                                                 young_person_df.columns[8]: 'timestamp',
                                                 young_person_df.columns[9]: 'timing'})
young_person_df=young_person_df.drop(young_person_df.index[[0,1]])
#確認
young_person_df.head(5)
```

話者ID、発話タイミングラベル、音声データのパスのみを抽出し、全てNanの行を排除

```python
young_person_df=young_person_df.loc[:,['ID','timing','音声']]
young_person_df=young_person_df.dropna(how='all')
young_person_df=young_person_df.reset_index()
young_person_df.head(5)
```

フォルダ作成

```python
import os

wav_full_path = output_folder + "/wav_ful"
# フォルダが既に存在するか確認
if not os.path.exists(wav_full_path):
# フォルダが存在しない場合、階層ごとにフォルダを作成
    os.makedirs(wav_full_path)
    print(f"フォルダを作成しました: {wav_full_path}")
else:
    print(f"フォルダは既に存在します: {wav_full_path}")

text_full_path = output_folder + "/text_ful"
# フォルダが既に存在するか確認
if not os.path.exists(text_full_path):
# フォルダが存在しない場合、階層ごとにフォルダを作成
    os.makedirs(text_full_path)
    print(f"フォルダを作成しました: {text_full_path}")
else:
    print(f"フォルダは既に存在します: {text_full_path}")
```

発話タイミングラベルをGoogedriveからダウンロード

```python
import subprocess
from tqdm import tqdm
import warnings

# UserWarningを無視する
warnings.filterwarnings("ignore", category=UserWarning)

# 成功と失敗を追跡するリスト
success_downloads = []
failed_downloads = []

for i in tqdm(range(len(young_person_df.ID)), desc="ダウンロード進捗"):
    speaker = young_person_df.ID[i]
    url = young_person_df.timing[i]
    url_ID = url.split("/")
    text_url_ID = url_ID[5]
    
    text_destination = text_full_path+"/"+speaker+"_full.txt"
    
    # gdownコマンドの構築
    command = f"gdown https://drive.google.com/uc?id={text_url_ID} -O {text_destination}"
        
    # コマンドの実行（標準出力と標準エラー出力を抑制）
    result = subprocess.run(command, shell=True)
    #, check=False, stdout=subprocess.PIPE, stderr=subprocess.PIPE
    

    # 結果に基づいてリストを更新
    if result.returncode == 0:
        success_downloads.append(text_destination)
    else:
        failed_downloads.append(text_destination)

# 警告のフィルタリングを元に戻す
warnings.filterwarnings("default", category=UserWarning)

# 成功と失敗の結果を表示
print(f"\nダウンロード成功: {len(success_downloads)}件")
#for file in success_downloads:
    #print(f"成功: {file}")

print(f"\nダウンロード失敗: {len(failed_downloads)}件")
for file in failed_downloads:
    print(f"失敗: {file}")
```

# **フル尺の音声ファイルをGoogledriveからダウンロード**　
##### (アクセス制限がかかるとエラーが発生する。少し時間を待ってもう一度試そう)

```python
import os
import subprocess
from tqdm import tqdm
import warnings
import sys

success_list_wav = output_folder + "/success_list_wav.txt"

# 成功したダウンロードのリストを保存する関数
def save_success_list(success_downloads):
    with open(success_list_wav, 'w') as f:
        for item in success_downloads:
            f.write("%s\n" % item)

# 成功したダウンロードのリストを読み込む関数
def load_success_list():
    try:
        with open(success_list_wav, 'r') as f:
            return [line.strip() for line in f]
    except FileNotFoundError:
        print("Fileが見つかりません")
        return []



# 既にダウンロードされたファイルのリストを読み込む
success_downloads = load_success_list()

# UserWarningを無視する
warnings.filterwarnings("ignore", category=UserWarning)

failed_downloads = []

for i in tqdm(range(len(young_person_df.ID)), desc="ダウンロード進捗"):
    speaker = young_person_df.ID[i]
    wav_destination = f"{wav_full_path}/{speaker}_full.wav"
    
    if wav_destination in success_downloads:
        print(f"{wav_destination} は既にダウンロードされています。スキップします。")
        continue
    
    url = young_person_df.音声[i]
    url_ID = url.split("/")[-2]  # URLからIDを抽出
    command = f"gdown https://drive.google.com/uc?id={url_ID} -O {wav_destination}"
    
    result = subprocess.run(command, shell=True, check=False, stdout=subprocess.PIPE, stderr=subprocess.PIPE)
    
    if result.returncode == 0:
        success_downloads.append(wav_destination)
    else:
        print(f"ダウンロード失敗: {wav_destination}")
        print("エラーメッセージ:", result.stderr.decode('utf-8'))
        save_success_list(success_downloads)  # 現在の成功リストを保存
        sys.exit(f"エラーによりプログラムを終了します。(時間を経過させて再度実行してください)")

# 警告のフィルタリングを元に戻す設定
warnings.filterwarnings("default", category=UserWarning)

# 成功と失敗の結果を表示
print(f"\nダウンロード成功: {len(success_downloads)}件")
print(f"\nダウンロード失敗: {len(failed_downloads)}件")

# 成功したダウンロードのリストを保存
save_success_list(success_downloads)
```
